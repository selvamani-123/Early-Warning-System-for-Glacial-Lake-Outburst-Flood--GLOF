# GLOF Sentinel - Model Evaluation Report

This notebook evaluates the performance of the Random Forest model trained to predict Glacial Lake Outburst Flood (GLOF) risk levels based on environmental factors.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

%matplotlib inline
```

## Load the Dataset and Model

```python
df = pd.read_csv('training/historical_glof_data.csv')
print(f"Loaded dataset with {len(df)} records.")

# Assuming labels and features have been applied
from training.feature_engineering import apply_feature_engineering
from training.label_generation import generate_risk_labels

df = apply_feature_engineering(df)
df = generate_risk_labels(df)
print("Risk Distribution:")
print(df['risk'].value_counts())
```

## Feature Importance

```python
model = joblib.load('glof_model.pkl')
features = joblib.load('glof_model_features.pkl')

importances = model.feature_importances_
feature_imp = pd.Series(importances, index=features).sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=feature_imp, y=feature_imp.index)
plt.title('Feature Importances in GLOF Prediction')
plt.xlabel('Importance Score')
plt.ylabel('Features')
plt.tight_layout()
plt.show()
```
